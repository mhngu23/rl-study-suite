```python
import pandas as pd
import matplotlib.pyplot as plt

# Paths to your CSV files
reinforce_csv = "results/reinforce_cartpole_returns.csv"
baseline_csv = "results/reinforce_with_baseline_cartpole_returns.csv"
dqn_csv = "results/dqn.csv"

# Load CSVs
reinforce_df = pd.read_csv(reinforce_csv)
baseline_df = pd.read_csv(baseline_csv)
dqn_df = pd.read_csv(dqn_csv)

# Optional: smooth the returns using a rolling average for clarity
window = 10  # rolling window size
reinforce_df['smoothed_return'] = reinforce_df['return'].rolling(window).mean()
baseline_df['smoothed_return'] = baseline_df['return'].rolling(window).mean()
dqn_df['smoothed_return'] = dqn_df['return'].rolling(window).mean()

# Plot
plt.figure(figsize=(10,6))
plt.plot(reinforce_df['episode'], reinforce_df['smoothed_return'], label="REINFORCE")
plt.plot(baseline_df['episode'], baseline_df['smoothed_return'], label="REINFORCE + Baseline")
plt.plot(dqn_df['episode'], dqn_df['smoothed_return'], label="DQN")
plt.xlabel("Episode")
plt.ylabel("Return (smoothed)")
plt.title("Training Returns per Episode")
plt.legend()
plt.grid(True)
plt.show()
```
